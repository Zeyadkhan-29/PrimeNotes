```python
import json
```

```python
def load_data(filename):
    with open(filename,"r") as f:
        data=json.load(f)
    return data
```

```python
data=load_data("stored_data.json")
print(data)
print(type(data))
```

```python
#Clean rating
def clean_data(data):
    text_to_num={"one":1, "two":2, "three":3, "four":4, "five":5}
    cleaned_data=[]
    unique_data=set()
    for user in data:
        raw_rating=str(user["rating"]).strip().lower()
        if raw_rating in text_to_num:
            raw_rating=text_to_num[raw_rating]
        user["rating"]=raw_rating

        #missing data
        raw_age=user.get("age")
        if(raw_age==None):
            user["age"]=None;

        #deduplication
        if(user["name"].strip() in unique_data):
            continue
        unique_data.add(user["name"])
        cleaned_data.append(user)

    return cleaned_data
            
    
```

```python
data=clean_data(data)
```

```python
def get_insights(data):
    # average rating
    total_rating = 0

    for user in data:
        total_rating += float(user["rating"])

    average_rating = total_rating / len(data)
    print(f"avg rating = {average_rating}")

    # percentage of users with poor rating
    poor_ratings = 0

    for user in data:
        if float(user["rating"]) < 3:
            poor_ratings += 1

    print(f"% of user with poor rating = {poor_ratings / len(data) * 100}%")
```

```python
get_insights(data)
```

```python
def get_recommendation(data):
    recommendations=[]
    for user in data:
        curr_recomm ={}
        curr_recomm["name"]=user["name"]
        if(float(user["rating"])>=4):
            curr_recomm["brand"]=["Apple"]
        else:
            curr_recomm["brand"]=["Samsung"]

        recommendations.append(curr_recomm)
    return recommendations        
```

```python
get_recommendation(data)
```

```python
#cleaning data - handling missing values

df=pd.read_csv("raw_data.csv")
df.isnull()
df.isna() #alies of isnull
df.isnull().sum()
df.dropna() #drop rows where NaN 
df.dropna(axis=1) #drop column where NaN 
df.fillna(0)

#filling column nall values with mean/avg values
age_mean=df["age"].mean()
cleaned_data=df.copy()
age_mean=cleaned_data["age"].mean()
cleaned_data["age"]=cleaned_data["age"].fillna(age_mean)
#cleaned_data

df.ffill()
df.bfill()
```

```python
#cleaning data - handling dublicate values
df.duplicated() #row wise
df["country"].duplicated()
df[["country","name"]].duplicated()
df2=df.copy 
df.drop_duplicates(inplace=True) #we should not use inplace we should make copy of original data and then clean
```

```python
#cleaning data - (handling data types and date-time)
df.dtypes
df2=df.copy()
df2=df2.fillna(0)
df2["age"]=df2["age"].astype("int64").copy()
df2.dtypes

df3=df.copy()
df3=df3.fillna(0)
df3["income"]=df3["income"].astype("int64").copy()
df3.dtypes
```

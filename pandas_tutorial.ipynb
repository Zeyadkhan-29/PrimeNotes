```python
#Series
s=pd.Series([1,2,3,4,"abc",5])
print(s)
print(type(s))
print(s.dtype)

#index
print(s[1])
print(s[0])
```

```python
#lebeling
s=pd.Series([19, 6, 14, 16], index=(["Zeyad", "Aly", "Wassam", "Waqas",]))
print(s)

print(s.Zeyad)
print(s["Zeyad"])
print(s.Aly)

print(s[0])
print(s[1])

print(s.index)
```

```python
#vectorization
s1=pd.Series([1,2,3,4,5])
s2=pd.Series([10,20,30,40,50])

print(s1+s2)

#handle missing values with (NaN)
#mutable values, immutable size
s1[0]=100
print(s1)

changed=s1.drop(0)
print(s1)
print(changed)
```

```python
#DataFrame
#(1) creating with the dictionary of list
info={
    "NAME":["Zeyad", "Aly", "Waqas", "Wassam"],
    "AGE":[19,5,16,14],
    "CGPA":[9.1,9.5,10,9.8]
}
df=pd.DataFrame(info)
print(df,type(df))

print(df.index)
print(df.columns)
```

```python
df
```

```python
#(2) creating with list of list

df=pd.DataFrame([["zeyad",19,9.1],["aly",5,9],["wassam",14,9.5],["waqas",16,10]], columns=["NAME","AGE","CGPA"])
print(df)
```

```python
#(3) creating with numpy arrays 
import numpy as np
np_arr=np.array([[1,2,3],[4,5,6]])
df=pd.DataFrame(np_arr, columns=["A","B","C"])
print(df)
```

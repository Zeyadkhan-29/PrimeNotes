```python
#transforming data (feature engineering)
import pandas as pd
df=pd.read_csv("raw_data.csv")
df2=df.copy()

#apply() method 
df2["tax"]=df2["income"].apply(lambda x : "20%" if x >= 50000 else "10%")

#map() method
changing = {"Male":"M", "Female":"F", "Unknown":"U",}
df2["gender"]=df2["gender"].map(changing)
df2

#assign () method

df2=df2.assign(new_income = (df2["income"] * 1.1).fillna(0))

#replace () method
df2["country"]=df2["country"].replace("USA","INDIA")
df2["country"].replace("USA","INDIA")
#df2["country"]=df2["country"].replace("NaN","INDIA") #its not work because that is empty place
df2["country"]=df2["country"].fillna("INDIA") #first fill empty space 
df2["country"]=df2["country"].replace("INDIA","USA") #then replace
df2
```

```python
#data transforming(other methods)

#to change the name of columns -> rename() method
df2=df2.rename(columns={"income":"salary","new_income":"new_salary"})

#to change lable of rows
df2=df2.rename(index={1:"First",2:"Second"})

#sort_values() method
df2["salary"].sort_values()
df2.sort_values("age")
df2["age"]=df2["age"].fillna(50)
df2.sort_values("salary",ascending=False)
sorted_data=df2.sort_values(["salary","new_salary"])
df2

#sort_index() method & #reset_index() method
sorted_data = sorted_data.reset_index(drop=True)
sorted_data.sort_index()
sorted_data.reset_index()

#rand() method
sorted_data["ranking"]=sorted_data["new_salary"].rank(ascending=False,method="dense") #to resole ties we method="dense"
#reorder columns
sorted_data[["id","name","age","country","gender","salary","tax","ranking","new_salary"]]
```

```python
# Melt & Pivot (for reshaphing)
import pandas as pd
df = pd.DataFrame({
    "country": ["USA", "USA", "India", "India"],
    "year": [2020, 2021, 2020, 2021],
    "sales": [100, 120, 90, 110],
    "profit": [20, 25, 18, 22]
})

melted_df=df.melt(
    id_vars=["country","year"],
    value_vars=["sales","profit"],
    var_name="matrix",
    value_name="value"

)

melted_df.pivot(
    index=["country","year"],
    columns="matrix",
    values="value"
)
```

```python
# Merging & Joining Data

df_customers = pd.DataFrame({
    "customer_id": [1, 2, 3, 4],
    "name": ["Adam", "Bob", "Charlie", "Dave"]
})

df_orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104],
    "customer_id": [2, 1, 4, 5],
    "amount": [250, 120, 300, 180]
})

pd.merge(df_customers, df_orders, on="customer_id") #inner join
pd.merge(df_customers, df_orders, on="customer_id", how="left") #left join
pd.merge(df_customers,  df_orders, on="customer_id", how="right") #right join
pd.merge(df_customers, df_orders,  on="customer_id", how="outer") #outer join
```
